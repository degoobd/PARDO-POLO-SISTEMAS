Resumen: Aprendizaje no supervisado

El aprendizaje no supervisado trabaja con datos sin etiquetas y busca encontrar patrones o grupos por sí mismo. Se puede comparar con un pastel, porque es la base sobre la que después se pueden agregar detalles.

Clustering: agrupa datos que se parecen entre sí.
K-means: crea una cantidad determinada de grupos y asigna cada dato al grupo más cercano.
Segmentación de imágenes: agrupa píxeles similares para identificar regiones como personas, cielo o carretera.
DBSCAN: encuentra grupos con formas irregulares y también puede detectar datos aislados.
Detección de anomalías: identifica elementos que son diferentes a lo considerado normal.
GMM: calcula probabilidades para saber a qué grupo pertenece cada dato y ayuda a encontrar anomalías.
Aprendizaje activo: combina máquina y humano; la máquina pregunta al humano solo cuando tiene dudas, reduciendo el trabajo de etiquetar datos.

En conclusión: el aprendizaje no supervisado permite convertir datos desordenados en grupos, patrones y señales útiles para tomar decisiones y descubrir información que no se conocía de antemano.
